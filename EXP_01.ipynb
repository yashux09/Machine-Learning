```python
import pandas as pd


df = pd.read_csv("/content/used_phone_price_prediction_1M.csv")

df = df.drop(columns=["warranty_remaining_months", "charger_available", "box_available","screen_size_inches","processor_score","repair_history","screen_cracked","city_tier","market_demand_score","screen_cracked","water_damage"])


print("First 5 Rows")
print(df.head())

print("\nLast 5 Rows")
print(df.tail())

print("\nShape of Dataset")
print(df.shape)


print("\nSize of datadet")
print(df.size)

print("\nColumn Names")
print(df.columns)


print("\nDataset Information")
print(df.info())

print("\nDuplicate Rows")
print(df.duplicated().sum())


df = df.drop_duplicates()

print("\nHighest Price Phones")
print(df.sort_values(by="resale_price", ascending=False).head())

print("\nLowest Price Phones")
print(df.sort_values(by="resale_price").head())


print("\nPhones Above ₹20,000")
print(df[df["resale_price"] > 20000].head())

df

```
